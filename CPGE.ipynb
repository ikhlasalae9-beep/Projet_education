### Cette dataset de source https://www.data.gov.ma/fr  recense la liste des établissements publics abritant des Classes Préparatoires aux Grandes Écoles (CPGE) au Maroc (mise à jour de juillet 2026). 
### Il contient 32 établissements répartis à travers les différentes régions du Royaume. L'objectif principal de ce projet étant la visualisation

| Colonne | Type | Description / Signification |
| :--- | :--- | :--- |
| **`NOM_ETABL`** | Texte | Nom officiel de l'établissement (Lycée) en français. |
| **`NOM_ETABA`** | Texte | Nom de l'établissement en langue arabe. |
| **`AdresseL`** | Texte | Adresse complète de l'établissement en caractères latins (français). |
| **`AdresseA`** | Texte | Adresse complète de l'établissement en langue arabe. |
| **`LL_COM`** | Texte | Nom de la commune ou de la municipalité d'implantation. |
| **`LL_PROV`** | Texte | Province ou préfecture de rattachement de l'établissement. |
| **`LL_REG`** | Texte | Région administrative du Maroc où se situe l'établissement. |

<h2 style="color: #a8e1e9; border-bottom: 2px solid #3498db; padding-bottom: 5px; margin-top: 20px;">
   3. Analyse Exploratoire
</h2>

```python
dt['region'].value_counts()
```

```python
fig = px.bar(dt['region'].value_counts(), 
             x=dt['region'].value_counts().index, 
             y=dt['region'].value_counts().values, 
             color=dt['region'].value_counts().values, 
             labels={'x':'Region', 'y':'Count'}, 
             title='Distribution of CPGE by Region')

fig.update_layout(xaxis_title='le nom de Region ', 
                  yaxis_title='nombre d\'etablissements')

fig.show()
```

 ### on remarque que la region la plus dominantes c'est casablanca_settat par 6 etablissement, suiver par fes_meknes, beni mellal-khnifra et rabat-sale-kenitra par 4 poue chacune
